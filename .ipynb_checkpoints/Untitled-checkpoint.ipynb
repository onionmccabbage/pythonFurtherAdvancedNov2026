```python
# we may import libraries
import numpy as np # pip install numpy
```

```python
# Numeric Python (Numpy) is extremely performant (written in C)
# we have additonal data types
# Numpy adds array and arange types
# arange is an array-range
n = np.arange(2, 8, 0.5) # start, stop-before, step
n.dtype # NB array and arange may only contain ONE data type
n
```

```python
# we may build an array from lists or tuples
l = [3,4,5]
t = (9,8,7)
m = np.array([l, t])
m # we have a two-dimensional array, also known as matrices (a matrix)
```

```python
# as many dimensions as you like
e = np.random.randn(2,3,4,5,2,3) # 6-dimensional array
```

```python
# creating empty arrays
f = np.empty( (3,6) ) # we get an array populated with garbage
g = np.zeros( (3,6), 'float32' )
g # careful: we sometimes get non-zero entries
```

```python
h=np.empty( (3,6), 'int32' ) # we can force the emtpy to a data type (and therefore trully empty
h
```

```python
# we can change the data type of an array
s = f.astype(np.str_) # apply string type to every member of the array
s
```

```python
# applying oerations across an array
t = np.random.randn(9000,9000) # a large quantity of data
```

```python
t_updated = ((t+3)**3)
t_updated # this is extremely performant
```
