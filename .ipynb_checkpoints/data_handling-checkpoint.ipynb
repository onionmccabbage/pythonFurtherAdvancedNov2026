# Merging Data Sets, cleaning data and correlation

```python
import numpy as np
import pandas as pd
```

```python
# we will import two data sets and merge them
url_sp = 'https://raw.githubusercontent.com/onionmccabbage/pythonFurtherAdvancedNov2026/refs/heads/main/data/SP500.csv'
sp_df = pd.read_csv(url_sp)
url_aa = 'https://raw.githubusercontent.com/onionmccabbage/pythonFurtherAdvancedNov2026/refs/heads/main/data/aapl.csv'
aa_df = pd.read_csv(url_aa)
```

```python
sp_df # .head()
aa_df #.head()
```

```python
# we may need to work with actual dates (rather than str values)
# for sp we have '2013-09-30'
sp_df['Date'] = pd.to_datetime( sp_df['Date'], format='%Y-%m-%d' )  # replace str with dates
aa_df['Date'] = pd.to_datetime( aa_df['Date'], format='%Y-%m-%d' )  
aa_df.info()
```

```python
# concatenate these data
conc_df = pd.concat( [sp_df, aa_df] )
conc_df
```

```python
# or alternatively we may merge data sets
merged_df = sp_df.merge(aa_df, left_on=['Date'], right_on=['Date'], suffixes=('_sp', '_aa'))
```

```python
merged_df # if there were matching dates, they would be merged
```

#### Cleaning Data

```python
# Often we find missing values
d = ['A', 'B', np.nan, 'D']
y = pd.Series(d) # Series underlies hte structure of dataFrames
y.notnull()
```

```python
x = y.fillna('Q') # does not persist!!!
x
```

```python
# we will build a dataFrame
df = pd.DataFrame( np.random.randn(10, 3) ) # make a df from an array
# add a few missing members
df.iloc[0:4, 1] = np.nan
df.iloc[3,2] = np.nan
# techniques to handle missing data
m = df.mean()
m # the mean will ignore missing members
df.fillna( m ) # we may use a mean value to fill missing members
```
