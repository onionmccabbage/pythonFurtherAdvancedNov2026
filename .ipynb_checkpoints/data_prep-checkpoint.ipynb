## Preparing Data
#### mathematics, index, dates, missing data

```python
import numpy as np
import pandas as pd

# a class is class A():
# an instance is i = A() # make an instance of a class


```

```python
# load some data
url = 'https://raw.githubusercontent.com/onionmccabbage/pythonFurtherAdvancedNov2026/refs/heads/main/data/risk.csv'
risk_df = pd.read_csv(url) #, index_col='Country')
risk_df # pandas will always provide an underlying ordinal index 0, 1, 2....
# we may re-index to use existing data
# CAREFUL: many features of dataFrames are non-desctructive by design
# this is to prevent accidental loss of data
# risk_df.set_index('Country', inplace=True) # specify which column to use as the new index
risk_dfi = risk_df.set_index('Country') # instead of altering the original, make a copy
```

```python
# we may choose to use different headings for the columns
i = ['stpr', 'ltpr', 'ster','lter', 'ori', 'cir']
risk_dfi.columns = i
risk_dfi
```

#### Working with Index, loc and iloc

```python
# we may grab data with the index: use loc[] to locate records in our data
risk_dfi.loc['Paraguay'] # get just the row for this data set
risk_dfi.iloc[2] # there is ALWAYS an underlying ordinal index, accessed with iloc
```

```python
# this isuseful when we need a number of rows
risk_dfi.iloc[23:33:2] # start:stop-before:step
```

```python
# we may need to check for unique entries
risk_dfi['cir'].unique().shape # there are 100 unique entries in this column
risk_dfi.nunique() # unique members across all columns
```

```python
# sorting data
rs = risk_dfi.sort_index(ascending=False) # by default ascending is True
rs # NB if an operation is non-desctructive, it is a good idea to make a copy with the new structure
```

```python
# head and tail
rs.head(18)
rs.tail(22)
```

### Working with Dates

```python
url = 'https://raw.githubusercontent.com/onionmccabbage/pythonFurtherAdvancedNov2026/refs/heads/main/data/aapl.csv'
# when parsing dates, provide a list of columns names (or indeces) to be treated as dates
# if the date is non-stabdard, then provode a datetime strftime picture
aapl_df = pd.read_csv(url, parse_dates=['Date']) # we give it a hint as the the existing format of dates
aapl_df
# handle the dates after importing...
# aapl_df['Date'] = pd.to_datetime( aapl_df['Date'] )
```

```python
aapl_df.describe()
aapl_df.info()
```

### Handling Dates as we read data

```python
url = 'https://raw.githubusercontent.com/onionmccabbage/pythonFurtherAdvancedNov2026/refs/heads/main/data/SP500.csv'
sp_df = pd.read_csv(url) #, parse_dates=['Date']) #, date_format='%m/%d/%y') # only need a datetime picture for non-standard formats
# we may parse dates after reading in the data
# here we replace the existing 'Date' columns with a newly calculated 'Date' column
sp_df['Date'] = pd.to_datetime( sp_df['Date'], format='%m/%d/%y' ) # the format is what is in the EXISTING data
sp_df.info()
```
