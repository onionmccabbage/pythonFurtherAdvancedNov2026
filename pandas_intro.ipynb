### Using Pandas

```python
# we usually import stuff at the top
import numpy as np
import pandas as pd # we may need to pip install pandas
```

```python
# Pandas adds yet more data structures
# we now have range and dataFrame types
# dataFrame is very like a spreadsheet
# each column in a dataFrame is a range (containing just one data type)
# we may load up data from anywhere: excel, csv, json, xml, html etc...
url = 'https://raw.githubusercontent.com/onionmccabbage/pythonFurtherAdvancedNov2026/refs/heads/main/data/Salaries.csv'
# we can create a dataFrae from any source
salary_df = pd.read_csv(url)
# salary_df = pd.read_csv('data/Salaries.csv') # load from local location
salary_df
```

```python
# basic analysis of this data
salary_df.describe()
```

```python
# we may use autocomplete with the 'TAB' key
which_columns = ['discipline', 'salary', 'rank']
salary_df[ which_columns ] # two columns ( a list inside a selector)
```

### Grouping and Aggregating Data

```python
# we may gather related data together
df_rank = salary_df.groupby(['rank']) # we have a groupBy object
df_rank[['salary', 'service']].mean() # apply some analysis to our groupBy object
```

```python
# mini challenge
# show the mean salary for each rank, grouped by sex
whichColumns = ['rank', 'sex']
mean_df = salary_df.groupby(whichColumns)
mean_df['salary'].mean() # max, min, sum...
```

```python
# aggregating data
cols = ['phd','salary','service']
operations = ['mean', 'min','max']
salary_df[cols].agg(operations)
```
